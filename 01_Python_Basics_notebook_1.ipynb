## 8.1 if / elif / else

### Step 1 – Understand the Concept
Conditional statements allow a program to choose different actions depending on whether conditions are true or false.

### Step 2 – Demonstrate the Concept
A grading system can assign a result based on a student's mark. The program checks conditions from top to bottom and executes the matching branch.

### Step 3 – Implement the Concept


```python
mark = 78

if mark >= 90:
    grade = "A"
elif mark >= 75:
    grade = "B"
elif mark >= 50:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
```

**What I learned from this code**

The `if`, `elif`, and `else` branches represent alternative decisions. Only the first matching branch is executed.


## 8.2 for Loop

### Step 1 – Understand the Concept

### Step 3 – Implement the Concept


```python
students = ["Asha", "Rahul", "Vignesh"]

for student in students:
    print("Present:", student)
```

**What I learned from this code**

## 8.3 while Loop

### Step 1 – Understand the Concept
A `while` loop repeats a block as long as its condition remains true. It is useful when the number of repetitions depends on a changing condition.

### Step 2 – Demonstrate the Concept

### Step 3 – Implement the Concept


```python
attempts = 0

while attempts < 3:
    attempts += 1
    print("Attempt", attempts)
```

**What I learned from this code**

## 8.4 break

### Step 1 – Understand the Concept
`break` immediately exits the nearest loop. It is useful when the required result has already been found or continuing the loop is unnecessary.

### Step 2 – Demonstrate the Concept

### Step 3 – Implement the Concept

## 8.5 continue

## 10.5 Variable-Length Arguments — *args and **kwargs

## 10.6 Lambda Functions

```python
students = [
    {"name": "Asha", "score": 82},
    {"name": "Vignesh", "score": 91},
    {"name": "Rahul", "score": 76},
]

ordered = sorted(students, key=lambda student: student["score"], reverse=True)
print(ordered)
```

## 10.7 Recursive Functions

### Step 1 – Understand the Concept
A recursive function calls itself to solve a problem by reducing it into smaller versions of the same problem. A base case is essential to stop the recursion.

### Step 2 – Demonstrate the Concept
Calculating a factorial can be described as `n * factorial(n-1)` until the base case `1` is reached.

### Step 3 – Implement the Concept


```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

**What I learned from this code**

The function repeatedly calls itself with a smaller number. When `n` reaches `1`, the base case stops further calls.


# 11. Advanced Python

## 11.1 Iterators

### Step 1 – Understand the Concept
An iterator is an object that produces values one at a time. Python's iteration protocol uses `__iter__()` to obtain an iterator and `__next__()` to retrieve the next value.

### Step 2 – Demonstrate the Concept
Think of a queue of tickets: instead of handing over every ticket at once, the iterator gives one ticket at a time.

### Step 3 – Implement the Concept


```python
numbers = iter([10, 20, 30])

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

**What I learned from this code**

`iter()` creates an iterator, and each `next()` call retrieves the next value. After the final value, another `next()` would raise `StopIteration`.


## 11.2 Generators

### Step 1 – Understand the Concept
A generator is a convenient way to create an iterator. It produces values lazily, meaning values are generated when requested rather than all being stored at once.

### Step 2 – Demonstrate the Concept
A large report can be processed row by row instead of loading every row into memory simultaneously.

### Step 3 – Implement the Concept


```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1

for value in count_up_to(4):
    print(value)
```

**What I learned from this code**

Because the function contains `yield`, calling it returns a generator. Each iteration resumes execution from where it previously paused.


## 11.3 yield

### Step 1 – Understand the Concept
`yield` temporarily pauses a generator function and sends a value to the caller. When the next value is requested, execution resumes from that point.

### Step 2 – Demonstrate the Concept
It is similar to a vending machine: the machine gives one item, pauses, and waits for the next request.

### Step 3 – Implement the Concept


```python
def even_numbers(limit):
    for number in range(2, limit + 1, 2):
        yield number

generator = even_numbers(8)

print(next(generator))
print(next(generator))
print(next(generator))
```

**What I learned from this code**

Each `next()` request resumes the generator until the next `yield`. This avoids creating the entire sequence beforehand.


## 11.4 Decorators

### Step 1 – Understand the Concept
A decorator is a function that wraps another function to add behavior without changing the wrapped function's main code.

### Step 2 – Demonstrate the Concept
A security system can place an access check around an operation. The operation remains the same, but extra behavior happens before it runs.

### Step 3 – Implement the Concept


```python
def log_call(function):
    def wrapper():
        print("Function is starting")
        function()
        print("Function has finished")
    return wrapper

@log_call
def show_message():
    print("Hello from Python")

show_message()
```

**What I learned from this code**

`@log_call` replaces `show_message` with the wrapper returned by `log_call`. The wrapper adds messages before and after the original function.


## 11.5 Closures

### Step 1 – Understand the Concept
A closure occurs when an inner function remembers values from its enclosing function even after the outer function has finished executing.

### Step 2 – Demonstrate the Concept
Imagine creating personalized settings. Each generated function can remember the setting it was created with.

### Step 3 – Implement the Concept


```python
def make_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
print(triple(5))
```

**What I learned from this code**

The inner `multiply()` function remembers `factor`. The two returned functions therefore keep different values in their closures.


## 11.6 global and nonlocal

### Step 1 – Understand the Concept
`global` allows a function to rebind a module-level variable. `nonlocal` allows an inner function to rebind a variable belonging to an enclosing function. These should be used carefully because they introduce shared state.

### Step 2 – Demonstrate the Concept
A counter stored outside a function can be updated with `global`; a nested counter can be updated with `nonlocal`.

### Step 3 – Implement the Concept


```python
counter = 0

def increase_global():
    global counter
    counter += 1

increase_global()
print("Global counter:", counter)


def make_counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        return count

    return increase

counter_a = make_counter()
print(counter_a())
print(counter_a())
```

**What I learned from this code**

`global counter` changes the module-level variable. `nonlocal count` changes the variable belonging to the enclosing `make_counter()` function.

## 13.1 map()

### Step 1 – Understand the Concept

```python
prices = [100, 200, 300]

with_tax = list(map(lambda price: price * 1.18, prices))

print(with_tax)
```

## 13.9 id()

### Step 1 – Understand the Concept
`id()` returns an integer that identifies an object for its lifetime in the current Python implementation. It can help demonstrate object identity, but it should not normally be used as a persistent identifier.

### Step 2 – Demonstrate the Concept
Two variables may refer to the same list object. Comparing their identities demonstrates that they point to the same object.

### Step 3 – Implement the Concept


```python
first = ["Python", "C"]
second = first
third = ["Python", "C"]

print("first is second:", first is second)
print("first is third:", first is third)

print("id(first):", id(first))
print("id(second):", id(second))
print("id(third):", id(third))
```

**What I learned from this code**

`first` and `second` reference the same list, so `first is second` is `True` and their IDs match. `third` is a separate list with equal contents, so its identity is different.

The following mini-program combines variables, collections, a function, a loop, a condition, a comprehension, and built-in functions.

**Task:** Store student scores, calculate the average, identify passing students, and display the highest score.


```python
students = {
    "Asha": 82,
    "Rahul": 68,
    "Vignesh": 91,
    "Kiran": 74
}

def average_score(scores):
    return sum(scores) / len(scores)

scores = list(students.values())
average = average_score(scores)

passed = [name for name, score in students.items() if score >= 70]
highest = max(scores)

print("Average score:", round(average, 2))
print("Passed students:", passed)
print("Highest score:", highest)
print("Any score below 50?", any(score < 50 for score in scores))
print("All students scored at least 60?", all(score >= 60 for score in scores))
```
